```python
# Interaktiivne kontroll: loeme slaidid HTML-failist ja kuvame nimekirja
import os
from bs4 import BeautifulSoup

HTML_PATH = 'public/index.html'
if os.path.exists(HTML_PATH):
    with open(HTML_PATH, 'r', encoding='utf-8') as f:
        soup = BeautifulSoup(f.read(), 'html.parser')
    
    slides = soup.find_all('div', class_='slide')
    print(f"Esitluses on kokku {len(slides)} slaidi:\n")
    for i, s in enumerate(slides, 1):
        title_tag = s.find(['h1', 'h2'])
        title = title_tag.get_text().strip() if title_tag else "Pealkirjata slaid"
        print(f" [{i:02d}] {title}")
else:
    print(f"Viga: Faili {HTML_PATH} ei leitud.")
```
